# 02 Embedding 与 Word2Vec 小实验

目标：

1. 理解 `nn.Embedding` 是查表；
2. 看懂 `[B,L] → [B,L,E]`；
3. 用极小例子理解 Skip-Gram 正负样本。


````python
import torch
from torch import nn

torch.manual_seed(42)

vocab_size = 10
embedding_dim = 4

embedding = nn.Embedding(
    num_embeddings=vocab_size,
    embedding_dim=embedding_dim,
    padding_idx=0,
)

token_ids = torch.tensor([
    [2, 3, 4, 0],
    [5, 6, 0, 0],
])

vectors = embedding(token_ids)

print("token_ids:", token_ids.shape)
print("vectors:", vectors.shape)
vectors
````

## 1. Embedding 就是按行索引权重矩阵

````python
print("embedding.weight shape:", embedding.weight.shape)

token_id = 3

print(torch.allclose(
    embedding(torch.tensor(token_id)),
    embedding.weight[token_id],
))
````

````python
print("padding vector:")
print(embedding.weight[0])
````

## 3. Skip-Gram 的训练样本直觉

句子：

```text
i love deep learning
```

窗口大小为 1，中心词 `deep` 的正上下文是 `love` 和 `learning`。


````python
sentence = ["i", "love", "deep", "learning"]
window_size = 1

pairs = []

for center_index, center_word in enumerate(sentence):
    left = max(0, center_index - window_size)
    right = min(len(sentence), center_index + window_size + 1)

    for context_index in range(left, right):
        if context_index == center_index:
            continue

        pairs.append(
            (center_word, sentence[context_index])
        )

pairs
````

Negative Sampling 会再为每个中心词抽取若干“不是真实上下文”的词，形成二分类正负样本。
