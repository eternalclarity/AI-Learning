```python
import collections
import jieba
```

```python
def count_str(s: str = "I love love you you too"):
    return collections.Counter(s.split())


counter = count_str()

print(counter)
print(counter.items())
for item in counter.items():
    print(f"{item[0]} : {item[1]}")

sorted_items = sorted(counter.items(),
                      key=lambda x: x[1],
                      reverse=True)
print(sorted_items)
```

```python
text = "我喜欢学习机器学习"
tokens = jieba.lcut(text)
print(tokens)
```

### zip() 用法

```python
names = ['Tom', 'Jack', 'Lucy']
scores = [90, 85, 95]

# 1.构造列表，字典
result = list(zip(names, scores))
dictionary = dict(zip(names, scores))
print(result)
print(dictionary)

# 2.最常见用法：同时遍历两个列表
for name, score in zip(names, scores):
    print(f"{name} : {score}")

# 3.解压
data = [
    ('Tom', 90),
    ('Jack', 85),
    ('Lucy', 95)
]
names_z, scores_z = zip(*data)
print(names_z)
print(scores_z)
```

```python
names = ['Tom', 'Jack', 'Lucy']

print(list(enumerate(names)))
print(dict(enumerate(names)))

for i, name in enumerate(names):
    print(f"{i} : {name}")
```
