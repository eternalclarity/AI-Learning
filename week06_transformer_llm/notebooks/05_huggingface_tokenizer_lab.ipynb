# 05 Hugging Face Tokenizer Lab

这个 Notebook 需要安装 `transformers` 且首次运行需要下载 tokenizer。

目标不是学习 Hugging Face API 本身，而是把 Week 05 自己实现的 token/id/padding/mask 与真实预训练模型 tokenizer 对应起来。

````python
from transformers import AutoTokenizer

tokenizer = AutoTokenizer.from_pretrained("bert-base-uncased", use_fast=True)
texts = [
    "I love transformers.",
    "This movie was much better than I expected, although the ending was a little predictable.",
]

batch = tokenizer(
    texts,
    padding=True,          # 当前 batch 最长
    truncation=True,
    max_length=128,
    return_tensors="pt",
)

for key, value in batch.items():
    print(key, value.shape)
````

````python
print(tokenizer.convert_ids_to_tokens(batch["input_ids"][0]))
print(batch["attention_mask"][0])
````
